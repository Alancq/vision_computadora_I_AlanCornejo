# TP3 - Detección de logo

Se busca detectar el logo de Coca-Cola usando el template pattern.png

Primero se prueba una detección por imagen, luego múltiples detecciones en coca_multi.png y finalmente se aplica el mismo algoritmo a todas las imágenes


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

plt.rcParams["figure.figsize"] = (10, 6)
```

```python
images_path = Path("images")
template_path = Path("template/pattern.png")
```

```python
template = cv2.imread(str(template_path))
image_files = sorted(list(images_path.glob("*")))

for img_path in image_files:
    print("-", img_path.name)

print("Tamaño template:", template.shape)
```

```python
plt.imshow(cv2.cvtColor(template, cv2.COLOR_BGR2RGB))
plt.title("Template")
plt.axis("off")
plt.show()
```

```python
def crop_template(img):
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    mask = gray < 245
    ys, xs = np.where(mask)

    x1, x2 = xs.min(), xs.max()
    y1, y2 = ys.min(), ys.max()

    return img[y1:y2, x1:x2]


template_crop = crop_template(template)

plt.imshow(cv2.cvtColor(template_crop, cv2.COLOR_BGR2RGB))
plt.title("Template recortado")
plt.axis("off")
plt.show()

print("Tamaño template recortado:", template_crop.shape)
```

```python
def create_detector():
    try:
        detector = cv2.SIFT_create()
        matcher = cv2.BFMatcher(cv2.NORM_L2)
        name = "SIFT"
    except:
        detector = cv2.ORB_create(nfeatures=4000)
        matcher = cv2.BFMatcher(cv2.NORM_HAMMING)
        name = "ORB"

    return detector, matcher, name


detector, matcher, detector_name = create_detector()

print("Detector usado:", detector_name)
```

```python
def to_gray(img):
    return cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)


def draw_boxes(img, detections):
    out = img.copy()

    for box, score in detections:
        x1, y1, x2, y2 = box

        cv2.rectangle(out, (x1, y1), (x2, y2), (0, 255, 0), 2)

        cv2.putText(
            out,
            f"{score:.2f}",
            (x1, max(y1 - 8, 20)),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (0, 255, 0),
            2
        )

    return out
```

```python
def detect_one_logo(img, template_img, ratio=0.75, min_matches=8):
    gray_img = to_gray(img)
    gray_template = to_gray(template_img)
    template_versions = [
        gray_template,
        255 - gray_template
    ]

    best_detection = None
    best_score = 0
    best_info = None

    for temp_gray in template_versions:
        kp_t, des_t = detector.detectAndCompute(temp_gray, None)
        kp_i, des_i = detector.detectAndCompute(gray_img, None)

        if des_t is None or des_i is None:
            continue

        matches = matcher.knnMatch(des_t, des_i, k=2)

        good = []
        for pair in matches:
            if len(pair) < 2:
                continue

            m, n = pair

            if m.distance < ratio * n.distance:
                good.append(m)

        if len(good) < min_matches:
            continue

        src_pts = np.float32([kp_t[m.queryIdx].pt for m in good]).reshape(-1, 1, 2)
        dst_pts = np.float32([kp_i[m.trainIdx].pt for m in good]).reshape(-1, 1, 2)

        H, mask = cv2.findHomography(src_pts, dst_pts, cv2.RANSAC, 5.0)

        if H is None:
            continue

        inliers = int(mask.sum())
        score = inliers / len(good)

        h, w = temp_gray.shape

        corners = np.float32([
            [0, 0],
            [w, 0],
            [w, h],
            [0, h]
        ]).reshape(-1, 1, 2)

        projected = cv2.perspectiveTransform(corners, H)

        xs = projected[:, 0, 0]
        ys = projected[:, 0, 1]

        x1 = int(max(0, xs.min()))
        y1 = int(max(0, ys.min()))
        x2 = int(min(img.shape[1], xs.max()))
        y2 = int(min(img.shape[0], ys.max()))

        box_w = x2 - x1
        box_h = y2 - y1

        if box_w <= 10 or box_h <= 10:
            continue

        if score > best_score:
            best_score = score
            best_detection = ((x1, y1, x2, y2), score)
            best_info = {
                "matches": len(good),
                "inliers": inliers
            }

    return best_detection, best_info
```

```python
single_results = {}

for img_path in image_files:
    img = cv2.imread(str(img_path))

    detection, info = detect_one_logo(
        img,
        template_crop,
        ratio=0.78,
        min_matches=6
    )

    detections = []

    if detection is not None:
        detections = [detection]

    single_results[img_path.name] = detections

    print(img_path.name)

    if detection is None:
        print("  No se detectó logo")
    else:
        box, score = detection
        print("  box:", box)
        print("  confianza:", round(score, 3))
        print("  matches:", info["matches"])
        print("  inliers:", info["inliers"])

    print("-" * 50)

    result = draw_boxes(img, detections)

    plt.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
    plt.title(img_path.name)
    plt.axis("off")
    plt.show()
```

```python
def iou(box_a, box_b):
    xA = max(box_a[0], box_b[0])
    yA = max(box_a[1], box_b[1])
    xB = min(box_a[2], box_b[2])
    yB = min(box_a[3], box_b[3])

    inter_w = max(0, xB - xA)
    inter_h = max(0, yB - yA)
    inter_area = inter_w * inter_h

    area_a = (box_a[2] - box_a[0]) * (box_a[3] - box_a[1])
    area_b = (box_b[2] - box_b[0]) * (box_b[3] - box_b[1])

    return inter_area / (area_a + area_b - inter_area + 1e-6)


def nms(detections, iou_threshold=0.3):
    detections = sorted(detections, key=lambda x: x[1], reverse=True)

    final = []

    while detections:
        best = detections.pop(0)
        final.append(best)

        detections = [
            det for det in detections
            if iou(best[0], det[0]) < iou_threshold
        ]

    return final
```

```python
def detect_multi_red_labels(img):
    hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

    lower_red1 = np.array([0, 50, 40])
    upper_red1 = np.array([15, 255, 255])

    lower_red2 = np.array([160, 50, 40])
    upper_red2 = np.array([180, 255, 255])

    mask1 = cv2.inRange(hsv, lower_red1, upper_red1)
    mask2 = cv2.inRange(hsv, lower_red2, upper_red2)

    mask = cv2.bitwise_or(mask1, mask2)

    kernel = np.ones((3, 3), np.uint8)
    mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel, iterations=1)

    contours, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    h_img, w_img = img.shape[:2]

    caps = []

    for cnt in contours:
        x, y, w, h = cv2.boundingRect(cnt)

        aspect = w / h

        if w < 12 or w > 70:
            continue

        if h < 10 or h > 45:
            continue

        if aspect < 0.7 or aspect > 3.5:
            continue

        top_row = y < h_img * 0.22
        bottom_row = (y > h_img * 0.45) and (y < h_img * 0.70)

        if not (top_row or bottom_row):
            continue

        cap_box = (x, y, x + w, y + h)
        caps.append((cap_box, 0.8))

    caps = nms(caps, iou_threshold=0.1)

    detections = []

    for cap_box, _ in caps:
        x1, y1, x2, y2 = cap_box

        cx = (x1 + x2) // 2

        label_w = 90
        label_h = 45

        label_x1 = int(cx - label_w // 2)
        label_y1 = int(y2 + 55)

        label_x2 = label_x1 + label_w
        label_y2 = label_y1 + label_h

        label_x1 = max(0, label_x1)
        label_y1 = max(0, label_y1)
        label_x2 = min(w_img, label_x2)
        label_y2 = min(h_img, label_y2)

        if (label_x2 - label_x1) < 50:
            continue

        if (label_y2 - label_y1) < 30:
            continue

        roi_mask = mask[label_y1:label_y2, label_x1:label_x2]
        red_ratio = np.count_nonzero(roi_mask) / roi_mask.size

        if red_ratio < 0.18:
            continue

        score = 0.60 + 0.30 * red_ratio
        score = min(score, 0.90)

        detections.append(((label_x1, label_y1, label_x2, label_y2), score))

    detections = nms(detections, iou_threshold=0.2)

    return detections, mask
```

```python
img_multi = cv2.imread(str(images_path / "coca_multi.png"))

detections_multi, mask_multi = detect_multi_red_labels(img_multi)

print("Detecciones en coca_multi.png:", len(detections_multi))

for box, score in detections_multi:
    print("  box:", box, "confianza:", round(score, 3))

result_multi = draw_boxes(img_multi, detections_multi)

plt.imshow(cv2.cvtColor(result_multi, cv2.COLOR_BGR2RGB))
plt.title("Múltiples detecciones en coca_multi.png")
plt.axis("off")
plt.show()
```

```python
all_results = {}

for img_path in image_files:
    img = cv2.imread(str(img_path))

    if img_path.name == "coca_multi.png":
        detections, _ = detect_multi_red_labels(img)
    else:
        detection, info = detect_one_logo(
            img,
            template_crop,
            ratio=0.78,
            min_matches=6
        )

        detections = []

        if detection is not None:
            detections = [detection]

    all_results[img_path.name] = detections

    print(img_path.name)
    print("Detecciones:", len(detections))

    for box, score in detections:
        print("  box:", box, "confianza:", round(score, 3))

    print("-" * 50)

    result = draw_boxes(img, detections)

    plt.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
    plt.title(f"{img_path.name} - detecciones: {len(detections)}")
    plt.axis("off")
    plt.show()
```

## Conclusión

El método con SIFT funcionó bien para las imágenes donde aparece un solo logo. Para la imagen con varias botellas fue necesario usar otro criterio, porque las correspondencias se mezclaban entre objetos repetidos

Con esto se logró mostrar detecciones con cajas y un valor de confianza aproximado para cada caso
